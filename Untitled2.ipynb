```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("dataframe_basics").master("local[*]").getOrCreate()
```

## 1. Tworzenie DataFrame

```python
from datetime import date

data = [
    (1, "Alex", "Luna-1", "USA", 18, 1.245, date(2026, 4, 10), True)
]
```

## Schema
Spark sam zgaduje typy danych (schema inference), aby zdefiniować je ręcznie używamy structtype oraz structfield.

```python
columns = [
    "id",
    "astronaut",
    "mission",
    "country",
    "duration_days",
    "distnce_million_km",
    "start_date",
    "successful"
]
```

```python
df1 = spark.createDataFrame(data, columns)
```

```python
df1.printSchema()
```

# Wykres - dystans a czas trwania misji


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Konwersja ramki Spark (df1) do ramki Pandas
df_pd = df1.toPandas()

# Wykres punktowy: zależność pokonanego dystansu od czasu trwania misji
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df_pd,
    x="duration_days",
    y="distnce_million_km",
    hue="successful",
    s=100,
)

plt.title("Dystans a czas trwania misji")
plt.xlabel("Czas trwania (dni)")
plt.ylabel("Dystans (mln km)")
plt.grid(True, linestyle="--", alpha=0.7)

plt.show()
```

# Podstawowe typy danych (wybrane); służą do definiowania typów danych które znajdują się w ramkach danych.
* StringType: Odpowiednik str w Pythonie – służy do przechowywania tekstu (np. imiona, nazwy misji, kraje).  
* BooleanType: Odpowiednik bool – wartości logiczne True lub False (np. czy misja się udała).   
* IntegerType: 32-bitowa liczba całkowita z zakresem od ok. -2 miliardów do +2 miliardów (np. wiek, liczba dni).   
* LongType: 64-bitowa liczba całkowita – używana do bardzo dużych liczb całkowitych, gdy IntegerType nie wystarcza (np. unikalne ID w Big Data, liczby w miliardach).   
* FloatType: 32-bitowa liczba zmiennoprzecinkowa o pojedynczej precyzji (mniej dokładna, zajmuje mniej pamięci).   
* DoubleType: 64-bitowa liczba zmiennoprzecinkowa o podwójnej precyzji – domyślny typ dla liczb z przecinkiem w PySparku (np. dystans w mln km).
* DateType: Odpowiednik datetime.date w Pythonie – przechowywanie samej daty w formacie YYYY-MM-DD.
